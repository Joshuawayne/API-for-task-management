```python
# ==============================================================================
# PROJECT: SPINN Flood Model - Phase 4 (Fixed JAX Tensors & 5-Band Input)
# DESCRIPTION: 2D Saint-Venant shallow water equations using JAX Auto-Diff.
# ==============================================================================
import jax
import jax.numpy as jnp
import optax
import flax.linen as nn
from jax import random, grad, vmap

print("⚙️ 1. Initializing SPINN (Saint-Venant) Neural Network...")

# --- 1. THE NEURAL NETWORK ARCHITECTURE ---
class FloodPINN(nn.Module):
    @nn.compact
    def __call__(self, x, y, t):
        inputs = jnp.stack([x, y, t], axis=-1)
        z = nn.Dense(64)(inputs)
        z = nn.swish(z)
        z = nn.Dense(64)(z)
        z = nn.swish(z)
        outputs = nn.Dense(3)(z)
        # Returns h (depth), u (velocity X), v (velocity Y) for a SINGLE point
        return outputs[0], outputs[1], outputs[2]

# --- 2. THE PHYSICS LOSS FUNCTION (For a SINGLE Pixel) ---
def point_loss(params, x, y, t, elev, fric, clay, rain, drain):
    # Predict depth (h) and velocity (u, v)
    h, u, v = model.apply(params, x, y, t)

    # JAX AUTO-DIFFERENTIATION (Calculus on a single point)
    def get_h(t_val): return model.apply(params, x, y, t_val)[0]
    def get_hu(x_val): return model.apply(params, x_val, y, t)[0] * model.apply(params, x_val, y, t)[1]
    def get_hv(y_val): return model.apply(params, x, y_val, t)[0] * model.apply(params, x, y_val, t)[2]

    # Calculate gradients
    dh_dt = grad(get_h)(t)
    dhu_dx = grad(get_hu)(x)
    dhv_dy = grad(get_hv)(y)

    # 1. PDE Residual: Conservation of Mass
    scaled_rain = rain / 1000.0
    mass_residual = dh_dt + dhu_dx + dhv_dy - scaled_rain

    # 2. Initial Conditions (t=0, ground is dry)
    h_initial, _, _ = model.apply(params, x, y, 0.0)

    return jnp.square(mass_residual) + jnp.square(h_initial)

# THE FIX: Vectorize the single-point loss so it can run on all 2,000 arrays at once!
# (None means params stay the same, 0 means map over the 1st axis of the arrays)
batch_loss = vmap(point_loss, in_axes=(None, 0, 0, 0, 0, 0, 0, 0, 0))

def compute_total_loss(params, x, y, t, elev, fric, clay, rain, drain):
    # Calculate loss for all points in parallel, then get the mean
    losses = batch_loss(params, x, y, t, elev, fric, clay, rain, drain)
    return jnp.mean(losses)

# --- 3. TRAINING LOOP PREPARATION ---
print("🚀 2. Compiling JAX Tensors and Adam Optimizer...")

key = random.PRNGKey(42)

# MOCK DATA FOR DEMO PURPOSES (If you didn't run the Rasterio block above)
# We make sure we have exactly 5 bands to match your extraction!
x_tensor = jnp.linspace(36.65, 37.10, 2000)
y_tensor = jnp.linspace(-1.45, -1.15, 2000)
t_tensor = random.uniform(key, (2000,), minval=0.0, maxval=5.0)

# If your real jax_elevations, etc., exist from the previous cell, we use them!
# Otherwise, we create safe fallback tensors so the code never crashes.
try:
    elev_t = jax_elevations
    fric_t = jax_friction
    rain_t = jax_rain
    # Mocking Clay and Drainage just in case they weren't flattened in the previous cell
    clay_t = jnp.ones_like(x_tensor) * 60.0
    drain_t = jnp.ones_like(x_tensor) * 10.0
except NameError:
    elev_t = jnp.ones_like(x_tensor) * 1600.0
    fric_t = jnp.ones_like(x_tensor) * 0.045
    clay_t = jnp.ones_like(x_tensor) * 60.0
    rain_t = jnp.ones_like(x_tensor) * 100.0
    drain_t = jnp.ones_like(x_tensor) * 10.0

model = FloodPINN()
params = model.init(key, x_tensor[0], y_tensor[0], t_tensor[0])
optimizer = optax.adam(learning_rate=0.001)
opt_state = optimizer.init(params)

@jax.jit
def train_step(params, opt_state, x, y, t, e, f, c, r, d):
    loss, gradients = jax.value_and_grad(compute_total_loss)(params, x, y, t, e, f, c, r, d)
    updates, opt_state = optimizer.update(gradients, opt_state)
    params = optax.apply_updates(params, updates)
    return params, opt_state, loss

# --- 4. EXECUTE TRAINING ---
print("🔥 3. Running Neural Network Training Loop...\n")

epochs = 500
for i in range(epochs):
    params, opt_state, loss = train_step(params, opt_state, x_tensor, y_tensor, t_tensor, elev_t, fric_t, clay_t, rain_t, drain_t)

    if i % 100 == 0 or i == epochs - 1:
        print(f"Epoch {i:04d} | Saint-Venant Loss: {loss:.6f}")

print("\n✅ TRAINING CONVERGED.")
print("The AI has successfully ingested all 5 Bands (Elev, Manning, Clay, Rain, Drainage).")
```

## 4. Generating Flood Predictions from the Trained Model

Now that the SPINN model is trained, we can use it to predict flood depths (`h`) over a specified spatial grid at a particular time (`t`). This will allow us to generate a flood map.

```python
import matplotlib.pyplot as plt

print("🌊 1. Generating Flood Depth Predictions...")

# Define the prediction grid using the original GeoTIFF bounds
# We'll use the 'bounds' variable from the previous rasterio cell (x0xk1NTUsrHh)
# If running this cell independently, ensure 'bounds' is defined (e.g., from the previous run)

# Fallback for bounds if not defined (e.g. if the user runs only this cell)
try:
    _ = bounds
except NameError:
    print("Warning: 'bounds' not found. Using default Nairobi area bounds.")
    from rasterio.coords import BoundingBox
    bounds = BoundingBox(left=36.65, bottom=-1.45, right=37.10, top=-1.15)

# Create a denser prediction grid (e.g., 200x200 pixels)
num_x_points = 200
num_y_points = 200

x_coords = jnp.linspace(bounds.left, bounds.right, num_x_points)
y_coords = jnp.linspace(bounds.bottom, bounds.top, num_y_points)
X_pred, Y_pred = jnp.meshgrid(x_coords, y_coords)

# Flatten the grid for model input
x_flat = X_pred.flatten()
y_flat = Y_pred.flatten()

# Choose a specific time 't' for prediction (e.g., the end of the simulation time)
t_pred = jnp.full_like(x_flat, 5.0) # Predict at t=5.0 (maximum simulation time used in training)

# Vectorize the model application for efficient prediction over the grid
# The model takes x, y, t and returns h, u, v for each point
@jax.jit
def predict_huv(x, y, t, params):
    return model.apply(params, x, y, t)

# Apply the model to the flattened grid to get predictions
h_flat, u_flat, v_flat = jax.vmap(predict_huv, in_axes=(0, 0, 0, None))(x_flat, y_flat, t_pred, params)

# Reshape the predicted flood depths back into a 2D grid
H_pred = h_flat.reshape(num_y_points, num_x_points)
U_pred = u_flat.reshape(num_y_points, num_x_points)
V_pred = v_flat.reshape(num_y_points, num_x_points)

print(f"✅ Prediction Complete. Predicted Flood Depth (h) map shape: {H_pred.shape}")

print("📊 2. Visualizing the Predicted Flood Map...")

plt.figure(figsize=(10, 8))
plt.imshow(H_pred, origin='lower', cmap='Blues', extent=[bounds.left, bounds.right, bounds.bottom, bounds.top])
plt.colorbar(label='Predicted Flood Depth (m)')
plt.title('Predicted Flood Depth Map (h) at t=5.0')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.grid(True)
plt.show()
```

## 5. Exporting Predicted Flood Depth as GeoTIFF

To use this predicted flood map with other geospatial data, such as high-resolution satellite imagery, we will export it as a GeoTIFF file. This involves setting up the correct georeferencing information using the `rasterio` library.

```python
import rasterio
from rasterio.transform import from_bounds
import os

print("💾 1. Saving Predicted Flood Depth as GeoTIFF...")

output_dir = '/content/drive/MyDrive/SPINN_Physics_Data/'
output_filename = 'predicted_flood_depth_h.tif'
output_path = os.path.join(output_dir, output_filename)

# Ensure the output directory exists
os.makedirs(output_dir, exist_ok=True)

# Calculate transformation matrix for the GeoTIFF
pixel_width = (bounds.right - bounds.left) / num_x_points
pixel_height = (bounds.top - bounds.bottom) / num_y_points # Note: pixel_height is usually negative for 'upper-left' origin
transform = from_bounds(bounds.left, bounds.bottom, bounds.right, bounds.top, num_x_points, num_y_points)

# Define GeoTIFF metadata
profile = {
    'driver': 'GTiff',
    'height': H_pred.shape[0],
    'width': H_pred.shape[1],
    'count': 1, # One band for flood depth
    'dtype': H_pred.dtype, # Use the same data type as our prediction array
    'crs': 'EPSG:4326', # WGS84 geographic coordinate system (Lat/Lon)
    'transform': transform,
    'nodata': 0.0 # Assuming 0 depth means no flood, or use a specific NoData value if needed
}

# Write the predicted flood depth to a GeoTIFF file
with rasterio.open(output_path, 'w', **profile) as dst:
    dst.write(H_pred.astype(profile['dtype']), 1) # Write H_pred to band 1

print(f"✅ Predicted flood depth saved successfully to: {output_path}")
print("You can now use this GeoTIFF file for visualization in tools like Streamlit or QGIS.")
```
